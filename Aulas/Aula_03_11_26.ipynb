```python
import numpy as np
```

```python
y       = np.array([1, 2, 3, 4, 5])
y_pred  = np.array([1, 1, 3, 4, 5])
```

```python
from sklearn.metrics import root_mean_squared_error, mean_absolute_error, r2_score, mean_squared_error
```

```python
print(f"MAE.....: {mean_absolute_error(y, y_pred):.2f}")
print(f"MSE.....: {mean_squared_error(y, y_pred):.2f}")
print(f"RMSE....: {root_mean_squared_error(y, y_pred):.2f}")
print(f"R2......: {r2_score(y, y_pred):.2f}")
```

```python
from sklearn.datasets import load_diabetes
```

```python
X, y = load_diabetes(return_X_y=True)
```

```python
X.shape, y.shape
```

```python
y[0]
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.neighbors import KNeighborsRegressor
```

```python
modelo_linear = LinearRegression()
modelo_knn = KNeighborsRegressor()
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_treino, X_teste, y_treino, y_teste = train_test_split(X, y, random_state=100)
```

```python
# Treinar os modelos
modelo_linear.fit(X_treino, y_treino)

modelo_knn.fit(X_treino, y_treino)
```

## Predição sem ajuste de escala

```python
y_pred = modelo_linear.predict(X_teste)
```

```python
print("Regressão Linear")
print(f"MAE.....: {mean_absolute_error(y_teste, y_pred):.2f}")
print(f"MSE.....: {mean_squared_error(y_teste, y_pred):.2f}")
print(f"RMSE....: {root_mean_squared_error(y_teste, y_pred):.2f}")
print(f"R2......: {r2_score(y_teste, y_pred):.2f}")
```

```python
y_pred = modelo_knn.predict(X_teste)
```

```python
print("KNN Linear")
print(f"MAE.....: {mean_absolute_error(y_teste, y_pred):.2f}")
print(f"MSE.....: {mean_squared_error(y_teste, y_pred):.2f}")
print(f"RMSE....: {root_mean_squared_error(y_teste, y_pred):.2f}")
print(f"R2......: {r2_score(y_teste, y_pred):.2f}")
```

## Predição com ajuste de escala

```python
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
```

```python
# Pipeline linear e KNN
pipe_linear = make_pipeline(StandardScaler(), modelo_linear)
pipe_knn = make_pipeline(StandardScaler(), modelo_knn)
```

```python
# Treinar os modelos
pipe_linear.fit(X_treino, y_treino)
pipe_knn.fit(X_treino, y_treino)
```

```python
y_pred_linear = pipe_linear.predict(X_teste)
y_pred_knn = pipe_knn.predict(X_teste)
```

```python
print("Regressão Linear")
print(f"MAE.....: {mean_absolute_error(y_teste, y_pred_linear):.2f}")
print(f"MSE.....: {mean_squared_error(y_teste, y_pred_linear):.2f}")
print(f"RMSE....: {root_mean_squared_error(y_teste, y_pred_linear):.2f}")
print(f"R2......: {r2_score(y_teste, y_pred_linear):.2f}")

print("KNN Linear")
print(f"MAE.....: {mean_absolute_error(y_teste, y_pred_knn):.2f}")
print(f"MSE.....: {mean_squared_error(y_teste, y_pred_knn):.2f}")
print(f"RMSE....: {root_mean_squared_error(y_teste, y_pred_knn):.2f}")
print(f"R2......: {r2_score(y_teste, y_pred_knn):.2f}")
```
